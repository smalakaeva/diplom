```python
from opensearchpy import OpenSearch
import pandas as pd

client = OpenSearch(
    hosts=[{"host": "localhost", "port": 9200}],
    http_auth=("admin", "SamplePassword1!"),
    use_ssl=True,
    verify_certs=False,
    ssl_assert_hostname=False,
    ssl_show_warn=False
)

index_name = "student_predictions"

query = {
    "size": 10000,
    "query": {
        "match_all": {}
    }
}

response = client.search(index=index_name, body=query)

rows = [hit["_source"] for hit in response["hits"]["hits"]]

df = pd.DataFrame(rows)

print(df.head())
```

```python
import matplotlib.pyplot as plt

risk_counts = df["predicted_risk"].value_counts()

# мягкая синяя палитра
colors = ["#7fbfff", "#2f7fd6"]  # голубой + синий

plt.figure(figsize=(7, 7))
plt.pie(
    risk_counts,
    labels=risk_counts.index,
    autopct="%1.2f%%",
    colors=colors[:len(risk_counts)]
)

plt.title("Распределение true_risk")
plt.show()
```
